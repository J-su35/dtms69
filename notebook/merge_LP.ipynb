```python
import pandas as pd
```

```python
path1 = "C:/Code/dtms69/data/AMI/LP_oct2026.parquet"
path2 = "C:/Code/dtms69/data/AMI/LP_nov2026.parquet"
path3 = "C:/Code/dtms69/data/AMI/LP_dec2026.parquet"
```

```python
df_oct = pd.read_parquet(path1)
df_oct
```

```python
df_nov = pd.read_parquet(path2)
df_nov 
```

```python
df_dec = pd.read_parquet(path3)
df_dec
```

```python
df_merge = pd.concat([df_oct, df_nov, df_dec], axis=0)
```

```python
df_merge
```

```python
df_merge.isna().sum()
```

```python
df_merge.info()
```

```python
df_merge.describe()
```

```python
df_merge['kW_net'] = df_merge['kW average'] - df_merge['kW exp average']
```

```python
col = df_merge.pop('kW_net')
pos = df_merge.columns.get_loc('kW exp average')
df_merge.insert(pos + 1, 'kW_net', col)
```

```python
df_merge
```

```python
monthly_count = df_merge.groupby(pd.Grouper(freq='M'))['PEA_No'].nunique()

print(monthly_count)
```

```python
monthly_pea = df_merge.groupby(pd.Grouper(freq='ME'))['PEA_No'].unique()

print(monthly_pea.to_string())
```

```python
oct = set(df_merge[df_merge.index.month == 10]['PEA_No'].unique())
dec = set(df_merge[df_merge.index.month == 12]['PEA_No'].unique())

# มีใน Jan แต่ไม่มีใน Feb
missing_pea = oct - dec

print("Missing:", missing_pea)
```

```python
output_path = "E:/Documents/dtms69/merge_LP_2026.parquet"
```
